# **Scraping Ulasan Game HayDay — Google Play Store**
Mengambil ulasan aplikasi HayDay dari Google Play Store menggunakanlibrary google-play-scraper dengan filter bahasa Indonesia.Target: 12.000 ulasan terbaru.

```python
from google_play_scraper import reviews, Sort
import pandas as pd

# scraping review hayday dari play store
hasil, token = reviews(
    'com.supercell.hayday',
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=12000
)

df = pd.DataFrame(hasil)
print(len(df), "review terkumpul")
df.head()
```

```python
# ambil kolom yang dibutuhin
df = df[['userName', 'score', 'at', 'content']]

# buang review kosong dan duplikat
df = df.dropna(subset=['content'])
df = df.drop_duplicates(subset=['content'])
df = df.reset_index(drop=True)

print(len(df), "review setelah dibersihkan")
df['score'].value_counts().sort_index()

# simpan ke csv
df.to_csv('hayday_reviews.csv', index=False)
print("disimpan: hayday_reviews.csv")
df.head()
```
